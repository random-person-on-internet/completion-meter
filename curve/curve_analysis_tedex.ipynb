```python
import json
import os
import matplotlib.pyplot as plt

DATA_PATH = "./extracted_data/extracted_data_tedex_3.json"
SAVE_DIR = "graph_collector_style_tedex"
os.makedirs(SAVE_DIR, exist_ok=True)

# change when batching
START_INDEX = 0
END_INDEX = 3995 # IT IS NON INCLUSIVE

def get_collector_curve(items):
    seen = set()
    curve = []
    count = 0
    for item in items:
        if item not in seen:
            seen.add(item)
            count+=1
        curve.append(count)
    return curve

with open(DATA_PATH, "r", encoding="utf-8") as f:
    data = json.load(f)

sorted_keys = sorted(data.keys(), key=lambda x: int(x))
subset_keys = sorted_keys[START_INDEX:END_INDEX]

for file_name in subset_keys:
    content = data[file_name]
    combined_items = content.get("entities", []) + content.get("noun_phrases", [])
    if not combined_items:
        continue

    curve = get_collector_curve(combined_items)

    plt.figure(figsize=(6,3))
    plt.plot(curve, marker='o')
    plt.title(f"{file_name}")
    plt.xlabel("Position in List")
    plt.ylabel("Unique Count")
    plt.grid(True)
    plt.tight_layout()

    file_safe = file_name.replace("/","_").replace(".txt", "")
    save_path = os.path.join(SAVE_DIR, f"s{file_safe}_collector.png")
    plt.savefig(save_path)
    plt.close()

print(f"Generated {len(subset_keys)} collector graphs in {SAVE_DIR}")
```
