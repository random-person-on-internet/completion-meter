```python
import json
import numpy as np
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, r2_score
import joblib

with open("./chunk_clusters_labeled.json", "r", encoding="utf-8") as f:
    data = json.load(f)

X = np.array([d["embedding"] for d in data])
y = np.array([d["progress"] for d in data])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=42)

reg = RandomForestRegressor(n_estimators=100, random_state=42)
reg.fit(X_train, y_train)

y_pred = reg.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Absolute Error: {mae}")
print(f"R^2 Score: {r2}")

joblib.dump(reg, "./speech_progress_regressor.pkl")
print("Model saved to speech_progress_regressor.pkl")
```
