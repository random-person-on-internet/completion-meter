```python
import json
import numpy as np
from sklearn.decomposition import PCA
import hdbscan
import os
from collections import Counter
import umap

with open("./chunk_embeddings.json", "r", encoding="utf-8") as f:
    data = json.load(f)

embeddings = np.array([d["embedding"] for d in data])

# print("Reducing dimensionality with PCA:")
# pca = PCA(n_components=100)
# reduced_embeddings = pca.fit_transform(embeddings)
print("Reducing dimensionality with UMAP:")
reducer = umap.UMAP(n_components=20, random_state=42)
reduced_embeddings = reducer.fit_transform(embeddings)

print("Clustering with hdbscan:")
cluster = hdbscan.HDBSCAN(min_cluster_size=4, prediction_data=True)
cluster_labels = cluster.fit_predict(reduced_embeddings)

for d, label in zip(data, cluster_labels):
    d["cluster"] = int(label)

print(Counter(cluster_labels))

output_path = "./chunk_clusters.json"
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(data, f, indent=2)

print(f"Saved clustered chunks to {output_path}")
num_clusters = len(set(cluster_labels)) - (1 if -1 in cluster_labels else 0)
print(f"Identified {num_clusters} clusters")
```
