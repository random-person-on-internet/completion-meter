```python
import json
from collections import defaultdict

with open("./clusters/chunk_clusters_tedex.json", "r", encoding="utf-8") as f:
    chunks = json.load(f)

speech_chunks = defaultdict(list)
for chunk in chunks:
    speech_chunks[chunk["speech_id"]].append(chunk)

for speech_id, chunk_list in speech_chunks.items():
    chunk_list.sort(key=lambda x: x["position"])
    total = len(chunk_list) - 1 if len(chunk_list) > 1 else 1
    for i, chunk in enumerate(chunk_list):
        chunk["progress"] = round(i/total, 4)

updated_chunks = [chunk for chunks in speech_chunks.values() for chunk in chunks]

with open("./labelled/chunk_clusters_labeled_tedex.json", "w", encoding="utf-8") as f:
    json.dump(updated_chunks, f, indent=2)

print(f"Saved progress-labeled chunks to ./labelled/chunk_clusters_labeled_tedex.json")
```
