```python
import csv
import json
import ast

def csv_to_json(csv_file_path, json_file_path):
    with open(csv_file_path, mode="r", encoding="utf-8") as csv_f:
        csv_reader = csv.DictReader(csv_f)
        data = list(csv_reader)

    cleaned_data = []
    for entry in data:
        try:
            cleaned_entry={
                "talk_id": entry.get("talk_id", "").strip(),
                "title": entry.get("title", "").strip(),
                "speaker": entry.get("speaker_1", "").strip(),
                "transcript": entry.get("transcript", "").strip(),
                "description": entry.get("description", "").strip(),
            }
            topics_raw = entry.get("topics", "[]")
            try:
                cleaned_entry["topics"] = ast.literal_eval(topics_raw)
            except:
                cleaned_entry["topics"] = []

            duration_raw = entry.get("duration", "0")
            try:
                cleaned_entry["duration"] = int(duration_raw)
            except:
                cleaned_entry["duration"] = 0

            if len(cleaned_entry["transcript"].split()) < 50:
                continue

            cleaned_data.append(cleaned_entry)

        except Exception as e:
            print(f"Skipping entry due to error: {e}")

    with open(json_file_path, mode="w", encoding="utf-8") as json_f:
        json.dump(cleaned_data, json_f, indent=2)
        print(f"Json file created")

csv_path = "./ted_talks_en.csv"
json_path = "./ted_talks.json"

csv_to_json(csv_path, json_path)
```
